## 1.2 Carrito de compras
Desarrolla una clase Carrito que tenga un atributo de diccionario con los productos (clave: nombre del producto, valor: cantidad). Implementa metodos para agregar productos al carrito y para mostrar el contenido del carrito como una lista de tuplas

```python
class Carrito():
    def __init__(self, productos, cantidad):
        self.lista = {productos: cantidad}    

    def agregar_productos(self, algo, cantidad):
        self.lista[algo] = cantidad

    def mostrar(self):
        print(self.lista)

Carrito1 = Carrito("papitas", 42)
Carrito1.agregar_productos("coca", 679)
Carrito1.mostrar()

    

```
