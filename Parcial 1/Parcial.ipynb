```python
cadenita = 'asdasdaaaoioioi'

def contar_vocales(cadena):
    caracter_dic = dict()  # Guarda repeticion de caracteres en un diccionario
    for caracter in cadena: 
        if caracter in caracter_dic: # Si ya estaba en el dic() significa que se repite
            caracter_dic[caracter] += 1 # Continua el conteo
        else:
            if (caracter == "A" or caracter == "a" or
                caracter == "E" or caracter == "e" or
                caracter == "I" or caracter == "i" or
                caracter == "O" or caracter == "o" or
                caracter == "U" or caracter == "u"):
                    caracter_dic[caracter] = 1 # Si el caracter no esta en el diccionario, lo agrega
    print(caracter_dic)
contar_vocales(cadenita)
```

```python
listaA = ["procesadores", "whisky", "tazas", "monitores"]
listaB = ["tazas", "whisky", "cafe", "monitores"]

def interseccion(lista1, lista2):
    set_1 = set(lista1)
    set_2 = set(lista2)

    en_ambos = list(set_1 & set_2)

    print(en_ambos)

interseccion(listaA, listaB)
```

```python
lista = ["sol","luna","sol","hipo"]

def palabra_frecuente(lista):
    dic = {}
    for palabra in lista:
        dic[palabra] = dic.get(palabra, 0) + 1
    mas_frecuente = max(dic, key=dic.get)
    print(mas_frecuente)

palabra_frecuente(lista)
```
