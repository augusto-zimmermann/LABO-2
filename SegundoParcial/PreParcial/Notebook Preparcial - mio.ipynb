```python
# Clases hijas
class UsuarioBasico(Usuario):
	def __init__(self, nombre, edad):
		super().__init__(nombre, "lector", edad)

	def acciones_permitidas(self):
		return ["leer"]

class UsuarioEditor(Usuario):
	def __init__(self, nombre, edad):
		super().__init__(nombre, "editor", edad)

	def acciones_permitidas(self):
		return ["leer", "editar"]

class UsuarioAdmin(Usuario):
	def __init__(self, nombre, edad):
		super().__init__(nombre, "admin", edad)

	def acciones_permitidas(self):
		return ["leer", "editar", "eliminar"]

# Fábrica de instancias según el rol
def crear_usuario(diccionario):
	rol = diccionario["rol"]
	if rol == "admin":
		return UsuarioAdmin(diccionario["nombre"], diccionario["edad"])
	elif rol == "editor":
		return UsuarioEditor(diccionario["nombre"], diccionario["edad"])
	else:
		return UsuarioBasico(diccionario["nombre"], diccionario["edad"])

# Generar las instancias dinámicamente
instancias = [crear_usuario(u) for u in usuarios]

# Mostrar resultados
for usuario in instancias:
	print(usuario.presentarse(), "→", usuario.acciones_permitidas())
```

## Ejercicio 2 – Generadores y API de Clima

```python
import requests
import json

URL_METEO = (
	"https://api.open-meteo.com/v1/forecast?latitude=-34.61&longitude=-58.38&hourly=temperature_2m"
)

class ErrorConexion():
	pass

try:
	r = requests.get(URL_METEO, timeout=10)
	r.raise_for_status()
	data = r.json()
	
	with open("clima.json", "w") as f:
		json.dump(data, f, indent=4)
	
	print("Salio todo bien, el archivo quedo guardado")

except (requests.RequestException, ValueError) as e:
	print(f"flaco la re cagaste. No se que hiciste pero el error es {e}")
```

```python
def cargar_clima():
    try:
        with open("clima.json", "r") as read:
            contenido = json.load(read)
        return contenido
    except FileNotFoundError:
        print("Flaco no encontre el archivo, donde esta?")
        raise
    except json.JSONDecodeError as e:
        print(f"Error de decodificacion {e}")
        raise

clima = cargar_clima()
horas = clima["hourly"]["time"]
temperaturas = clima["hourly"]["temperature_2m"]
#print(horas)
#print(temperaturas)
print(list(zip(horas, temperaturas))) # de donde salio zip, no?
```

```python
def filtrar_calor(temperaturas, umbral):
    for temp in temperaturas:
        if temp > umbral:
            yield temp

gen = filtrar_calor(temperaturas, 25)
print(list(gen))
```

```python
redondear = lambda x: int(round(x))
dicc_10 = {
    horas[h]: redondear(temperaturas[h])
    for h in range(min(10, len(temperaturas)))
}
print("Hora > Temperaturas (10 primeras)")
for hora, temp in dicc_10.items():
    print(hora, ">", temp)
```

```python
from statistics import fmean

def formatear_salida(func):
    def wrapper(self):
        print("-" * 20)
        print("Reporte Climatico")
        print("-" * 20)
        func(self)
    return wrapper

class ReporteClimatico:
    def __init__(self, fecha, temperaturas):
        self.fecha=fecha
        self.temperaturas=temperaturas
    
    @formatear_salida
    def resumen(self):
        """
        Imprime temperatura maxima, minima y promedio.
        """
        tmax = max(self.temperaturas)
        tmin = min(self.temperaturas)
        tprom = fmean(self.temperaturas)
        print(f"Máxima:     {tmax:.2f} °C")
        print(f"Mínima:     {tmin:.2f} °C")
        print(f"Promedio:   {tprom:.2f} °C")
    
reporte = ReporteClimatico(temperaturas=temperaturas, fecha=hora)
salida = reporte.resumen()
```
