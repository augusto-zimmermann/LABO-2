## Ejercicio 1 – Gestion de Cursos y Decoradores (50 pts)

```python
cursos = [
    {"nombre": "Python Inicial", "nivel": "bsico", "alumnos": 35},
    {"nombre": "Machine Learning", "nivel": "avanzado", "alumnos": 22},
    {"nombre": "Bases de Datos", "nivel": "intermedio", "alumnos": 28},
    {"nombre": "Programacin Web", "nivel": "bsico", "alumnos": 40},
]

filtrado = list(filter(lambda cursos: cursos["nivel"]=="bsico", cursos))
nombres_bsico = [nombre["nombre"] for nombre in filtrado]
print(filtrado)
print(nombres_bsico)
```

```python
class Curso():
    def __init__(self, nombre, nivel, alumnos):
        self.nombre=nombre
        self.nivel=nivel
        self.alumnos=alumnos
    def info(self):
        return f"Curso: {self.nombre} (Nivel: {self.nivel}) - {self.alumnos} alumnos"

instancias = [
    Curso(
        nombre=curso["nombre"],
        nivel=curso["nivel"],
        alumnos=curso["alumnos"]
    )
    for curso in cursos
]

for alumno in instancias:
    print(alumno.info())
```
