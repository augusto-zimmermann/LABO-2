## Ejercicio 1 – Gestion de Vehiculos y Decoradores

### 1.

```python
vehiculos = [
	{"patente": "AB123CD", "tipo": "camion", "anio": 2018},
	{"patente": "AC555XY", "tipo": "auto", "anio": 2022},
	{"patente": "BB999AA", "tipo": "moto", "anio": 2019},
	{"patente": "CD321BA", "tipo": "camion", "anio": 2015},
]

camiones = list(filter(lambda x:x["tipo"]=="camion", vehiculos))
print("Vehiculos tipo camiones: ",camiones)
patentes_camiones = [camion["patente"] for camion in camiones]
print("Patentes de camiones: ", patentes_camiones)
```

### 2.

```python
class Vehiculo():
	def __init__(self, patente, tipo, anio):
		self.patente=patente
		self.tipo=tipo
		self.anio=anio
	def descripcion(self):
		return f"Vehiculo {self.patente} - {self.tipo} ({self.anio})"
	
instancias_vehiculo = [
	Vehiculo(
		patente=vehiculo["patente"],
		tipo=vehiculo["tipo"],
		anio=vehiculo["anio"]
	)
	for vehiculo in vehiculos
]

for vehiculo in instancias_vehiculo:
	print(vehiculo.descripcion())
```

### 3.

```python
class Auto(Vehiculo):
	def __init__(self, patente, anio):
		super().__init__(patente, "auto", anio)

	def info_extra(self):
		return "Este vehiculo tiene 5 plazas"

class Moto(Vehiculo):
	def __init__(self, patente, anio):
		super().__init__(patente, "moto", anio)

	def info_extra(self):
		return "Este vehiculo es agil en la ciudad"

class Camion(Vehiculo):
	def __init__(self, patente, anio):
		super().__init__(patente, "camion", anio)

	def info_extra(self):
		return "Este vehiculo tiene una capacidad de carga superior"

nuevos = [
	Auto('AAAAAA', 2016),
	Moto('BBBBBB', 2022),
	Camion('CCCCCC', 1994)
]

for v in nuevos:
	print(f"{v.patente}, {v.tipo}, {v.anio}")
	print(v.info_extra())
	print("\n")
```

### 4.

```python
class VehiculoAntiguoError():
	pass

def verificar_antiguedad(func):
	def wrapper():
		if getattr(vehiculo, "anio", None) > 2015:
			func()
		else:
			raise VehiculoAntiguoError(f"El vehiculo es muy antiguo maestro")
	return wrapper

@verificar_antiguedad
def registrar_vehiculo():
	print("Registrando vehiculo")
	
for vehiculo in instancias_vehiculo:
	try:
		registrar_vehiculo()
	except VehiculoAntiguoError as e:
		print(f"Error {e}")
```

### 5.

```python
def guardar_vehiculos():
	try:
		ruta_archivo = "vehiculos.txt"
		with open (ruta_archivo, "w") as f:
			for vehiculo in vehiculos:
				if isinstance(vehiculo, dict):
					patente, tipo, anio = vehiculo["patente"], vehiculo["tipo"], vehiculo["anio"]
				else:
					patente, tipo, anio = vehiculo.patente, vehiculo.tipo, vehiculo.anio
				f.write(f"{patente},{tipo},{anio}\n")
		print(f"Vehiculos guardados en {ruta_archivo}")
	except OSError as e:
		print(f"Error al escribir el archivo:  {e}")
		print("El archivo no fue guardado")

guardar_vehiculos()
```

### 2.

```python
def cargar_posiciones():
	try:
		with open("iss.json", "r") as read:
			contenido = json.load(read)
		return contenido
	except FileNotFoundError:
		print("Flaco no encontre el archivo, donde esta?")
		raise
	except json.JSONDecodeError as e:
		print(f"Error de decodificacion: {e}")
		raise

posiciones = cargar_posiciones()
latitudes = posiciones["iss_position"]["latitude"]
longitudes = posiciones["iss_position"]["longitude"]
timestamp = posiciones["timestamp"]

print(list(zip(latitudes, longitudes)))
```

### 3.

```python
def filtrar_latitud(posiciones, umbral):
	for latitud in posiciones:
		if latitud > umbral:
			yield latitud

umbral = 30
gen = filtrar_latitud(posiciones, umbral)
print(list(gen))
```

### 4.

```python
convertir_coord = lambda x:float(round(x, ndigits=2))
dicc_pos = {
	timestamp[t]: convertir_coord(latitudes[t])
	for t in range(nrequests)
}
for t, latitud in dicc_pos.items():
    print(t, ">", latitud)
```

### 5.

```python
from statistics import fmean

def encabezado_iss(func):
	def wrapper(self):
		print("=" * 10, "REPORTE DE POSICIONES DE LA ISS", "=" *10)
		func(self)
	return wrapper

class ReporteISS():
	def __init__(self, posiciones):
		#self.fecha=fecha
		self.posiciones=posiciones

	@encabezado_iss
	def resumen(self):
		lmax = max(self.posiciones)
		lmin = min(self.posiciones)
		lprom = fmean(self.posiciones)
		print(f"Latitud máxima:     {lmax:.2f} °C")
		print(f"Latitud mínima:     {lmin:.2f} °C")
		print(f"Latitud promedio:   {lprom:.2f} °C")

reporte = ReporteISS(posiciones=posiciones)
salida = reporte.resumen()
	
```
