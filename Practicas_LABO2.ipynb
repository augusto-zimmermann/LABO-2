## 1.5 Verificar palíndromo
Escribe un programa que verifique si una cadena es un palíndromo, es decir, si se lee igual de adelante hacia atrás como de atrás hacia adelante.


```python
# REVISAR

def es_palindromo(texto):
  """
  Comprueba si una cadena de texto es un palíndromo.

  Args:
    texto: La cadena de texto a verificar.

  Returns:
    True si la cadena es un palíndromo, False en caso contrario.
  """
  # 1. Convertir a minúsculas y eliminar espacios para estandarizar
  texto_procesado = ''.join(texto.lower().split())
  # 2. Comparar la cadena procesada con su versión invertida
  return texto_procesado == texto_procesado[::-1]

# Ejemplos de uso:
palabra1 = "madam"
palabra2 = "hola"
frase = "La ruta nos aportó otro paso natural"

print(f"'{palabra1}' es palíndromo: {es_palindromo(palabra1)}")
print(f"'{palabra2}' es palíndromo: {es_palindromo(palabra2)}")
print(f"'{frase}' es palíndromo: {es_palindromo(frase)}")
```

Implementa un script que realice operaciones con conjuntos para gestionar el inventario de una tienda. El script
utilizará tres conjuntos inventario_tienda1, inventario_tienda2 y productos_defectuosos, representando los
productos disponibles en dos tiendas y los productos defectuosos en ambas tiendas, respectivamente.
Ejemplo:
inventario_tienda1 = {"manzanas", "naranjas", "uvas", "platanos"}
inventario_tienda2 = {"peras", "uvas", "platanos", "kiwis"}
productos_defectuosos = {"uvas", "kiwis"}
El script debe realizar las siguientes operaciones:

● Encontrar la lista completa de productos disponibles en ambas tiendas (unión de inventario_tienda1 y
inventario_tienda2).

● Encontrar los productos que están disponibles en ambas tiendas (intersección de inventario_tienda1 y
inventario_tienda2).

● Encontrar los productos que están disponibles en cualquiera de las tiendas pero no en ambas (diferencia
simétrica entre inventario_tienda1 y inventario_tienda2).

● Encontrar los productos disponibles en ambas tiendas que no son defectuosos.

● Verificar si todos los productos defectuosos están en el inventario de la tienda 1 (subconjunto).

● El script debe mostrar los resultados de cada operación en la consola

```python
inventario_tienda1 = {"procesadores", "whisky", "tazas", "monitores"}
inventario_tienda2 = {"tazas", "whisky", "cafe", "monitores"}
productos_defectuosos = {"monitores", "tazas"}

productosDisponibles = inventario_tienda1 | inventario_tienda2 #union
print("Lista completa de productos:                                     ", productosDisponibles)

productosDisponiblesEnAmbasTiendas = inventario_tienda1 & inventario_tienda2 #interseccion
print("Productos diponibles en ambas tiendas:                           ", productosDisponiblesEnAmbasTiendas)

productosDisponiblesNoEnAmbasTiendas = inventario_tienda1 ^ inventario_tienda2 #diferencia
print("Productos no disponibles en ambas tiendas:                       ", productosDisponiblesNoEnAmbasTiendas)

productosDisponiblesEnAmbasTiendasNoDefectuosos = productosDisponiblesEnAmbasTiendas ^ productos_defectuosos #diferencia simetrica
print("Productos diponibles en ambas tiendas (no defectuosos):          ", productosDisponiblesEnAmbasTiendasNoDefectuosos)

print("Todos los productos defectuosos se encuentran en la tienda 1:    ", productos_defectuosos.issubset(inventario_tienda1)) #subconjunto
```
