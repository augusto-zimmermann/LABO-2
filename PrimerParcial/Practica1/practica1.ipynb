```python
def mostrar_info(**kwargs):
	for clave, valor in kwargs.items():
		print(f"{clave}: {valor}")

mostrar_info(nombre="Ana", edad=30, ciudad="Madrid")
# Output:
# nombre: Ana
# edad: 30
# ciudad: Madrid
```

```python
def sumar_todos(*args):
	return sum(args)

print(sumar_todos(1, 2, 3)) # Output: 6
print(sumar_todos(4, 5, 6, 7, 8)) # Output: 30
```
