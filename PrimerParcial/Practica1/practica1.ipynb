## 4. Interseccion y diferencia de conjuntos (8 pts)
Explicacion: Ejercicio para practicar operaciones basicas con conjuntos a partir de listas con duplicados.
- Consigna: Escribe `comparar_listas(A,B)` que devuelva una tupla (`en_ambos`, `solo_A`, `solo_B`) con elementos en comun y exclusivos.


```python
A = {"procesadores", "whisky", "tazas", "monitores"}
B = {"tazas", "whisky", "cafe", "monitores"}

def comparar_listas(A, B):
    set_A = set(A)
    set_B = set(B)
    
    en_ambos = set_A & set_B        # intersección
    solo_A = set_A - set_B          # diferencia
    solo_B = set_B - set_A          # diferencia
    
    print(f"En ambos: {en_ambos}. Exclusivo A: {solo_A}. Exclusivo B: {solo_B}")

comparar_listas(A, B)
```

## 5. Filtrar registros por condicion (10 pts)
Explicacion: Ejercicio de filtrado de diccionarios dentro de una lista, combinando ciclos, condicionales y manejo de claves inexistentes. 
- Consigna: Implementa `filtrar_registros(registros, campo, umbral)` que devuelva solo los registros con campo ≥ umbral.


```python
registros = [
    {"nombre": "Ana", "edad": 20},
    {"nombre": "Luis", "edad": 15},
    {"nombre": "Marta"},  # no tiene 'edad'
    {"nombre": "Pedro", "edad": 30},
]

def filtrar_registros(registros, campo, umbral):
    filtrados = []
    for registro in registros:
        if campo in registro and registro[campo] >= umbral:
            filtrados.append(registro)
    return filtrados

filtrar_registros(registros, campo="edad", umbral=19)
```
