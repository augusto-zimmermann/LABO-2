## 1. Ajuste de precios (8 pts)
Explicacion: Se pide procesar una lista de precios aplicando reglas distintas segun el rango de valores.
Es un ejercicio de ciclos y condicionales. 
- Consigna: Escribe una funcion `ajustar_precios(precios)` que devuelva una nueva lista ajustada segun:
  - Si precio < 100, aumentar 10%.
  - Si 100 ≤ precio < 300, dejar igual.
  - Si precio ≥ 300, reducir 5%.


```python
def ajustar_precios(precios):
    ajustados = []
    for p in precios:
        if p < 100:
            ajustados.append(round(p * 1.10, 2))
        elif p < 300:
            ajustados.append(p)
        else:
            ajustados.append(round(p * 0.95, 2))
    return ajustados
```

## 2. Limpieza de nombres (8 pts)
Explicacion: Ejercicio de normalizacion de datos y eliminacion de duplicados preservando el orden. 
- Consigna:
Implementa `limpiar_nombres(nombres)` que reciba una lista de strings, quite espacios, convierta a minúsculas y elimine duplicados.

```python
def limpiar_nombres(nombres):
    vistos = set()
    res = []
    for n in nombres:
        limpio = n.strip().lower()
        if limpio not in vistos:
            vistos.add(limpio)
            res.append(limpio)
    return res
```

## 4. Interseccion y diferencia de conjuntos (8 pts)
Explicacion: Ejercicio para practicar operaciones basicas con conjuntos a partir de listas con duplicados.
- Consigna: Escribe `comparar_listas(A,B)` que devuelva una tupla (`en_ambos`, `solo_A`, `solo_B`) con elementos en comun y exclusivos.


```python
A = {"procesadores", "whisky", "tazas", "monitores"}
B = {"tazas", "whisky", "cafe", "monitores"}

def comparar_listas(A, B):
    set_A = set(A)
    set_B = set(B)
    
    en_ambos = set_A & set_B        # intersección
    solo_A = set_A - set_B          # diferencia
    solo_B = set_B - set_A          # diferencia
    
    print(f"En ambos: {en_ambos}. Exclusivo A: {solo_A}. Exclusivo B: {solo_B}")

comparar_listas(A, B)
```

- Consigna: Implementa `filtrar_registros(registros, campo, umbral)` que devuelva solo los registros con campo ≥ umbral.


```python
registros = [
    {"nombre": "Ana", "edad": 20},
    {"nombre": "Luis", "edad": 15},
    {"nombre": "Marta"},  # no tiene 'edad'
    {"nombre": "Pedro", "edad": 30},
]

def filtrar_registros(registros, campo, umbral):
    filtrados = []
    for registro in registros:
        if campo in registro and registro[campo] >= umbral:
            filtrados.append(registro)
    return filtrados

filtrar_registros(registros, campo="edad", umbral=19)
```

## 6. Maximo y minimo con posiciones (10 pts)
Explicacion: Se busca trabajar con recorridos manuales de listas, sin usar funciones built-in de min o max.
- Consigna: Define `extremos(nums)` que retorne `(min val, pos min, max val, pos max)` con indices de la primera aparicion.


```python
def extremos(nums):
    if not nums:
        return None
    min_val, max_val = nums[0], nums[0]
    pos_min, pos_max = 0, 0
    for i, n in enumerate(nums):
        if n < min_val:
            min_val, pos_min = n, i
        if n > max_val:
            max_val, pos_max = n, i
    return min_val, pos_min, max_val, pos_max
```

## 7. Unir inventarios (10 pts)
Explicacion: Se necesita combinar inventarios parciales sumando cantidades por codigo. Ejercicio de acumulacion en diccionarios. 
- Consigna: Dadas dos listas de tuplas (codigo, cantidad), implementa `unir_inventarios(A,B)` que devuelva `{codigo: cantidad total}`.


```python
def unir_inventarios(A, B):
    inv = {}
    for cod, cant in A + B:
        inv[cod] = inv.get(cod, 0) + cant
    return inv
```

## 8. Ventanas contiguas validas (10 pts)
Explicacion: Se practica el concepto de ventanas deslizantes para optimizar calculo de sumas de sublistas.
Consigna: Implementa contar ventanas(nums, k, suma max) que cuente cuantas sublistas de tamaño k cumplen que su suma ≤ suma max.


```python
def contar_ventanas(nums, k, suma_max):
    count = 0
    for i in range(len(nums) - k + 1):
        if sum(nums[i:i+k]) <= suma_max:
            count += 1
    return count
```

## 9. Re-mapear etiquetas (10 pts)
Explicacion: Ejercicio de reemplazo condicionado usando un diccionario como mapa de traduccion. Consigna:
Implementa remapear etiquetas(datos, mapa) que reemplace la clave "etiqueta" de cada registro según el diccionario de mapeo.


```python
def remapear_etiquetas(datos, mapa):
    res = []
    for d in datos:
        nuevo = d.copy()
        if "etiqueta" in nuevo and nuevo["etiqueta"] in mapa:
            nuevo["etiqueta"] = mapa[nuevo["etiqueta"]]
        res.append(nuevo)
    return res
```

## 10. Tuplas de pares ordenados (8 pts)
Explicacion: Uso de tuplas y ciclos anidados con condicion sobre pares. Consigna: Escribe pares ordenados(nums)
que devuelva pares (a,b) con a < b y suma par, sin repetir pares.


```python
def pares_ordenados(nums):
    pares = []
    n = len(nums)
    for i in range(n):
        for j in range(i+1, n):
            a, b = nums[i], nums[j]
            if a < b and (a + b) % 2 == 0:
                pares.append((a, b))
            elif b < a and (a + b) % 2 == 0:
                pares.append((b, a))
    return pares
```

## 11. Palabras unicas y repetidas (6 pts)
Explicacion: Ejercicio clasico de conteo de frecuencias para separar palabras unicas y repetidas. Consigna:
Implementa analizar palabras(palabras) que retorne una tupla (unicas, repetidas) segun las frecuencias.


```python
def analizar_palabras(palabras):
    frec = {}
    for p in palabras:
        frec[p] = frec.get(p, 0) + 1
    unicas = {p for p, f in frec.items() if f == 1}
    repetidas = {p for p, f in frec.items() if f > 1}
    return unicas, repetidas
```

## 12. Compactar secuencias (8 pts)
Explicacion: Ejercicio para detectar repeticiones consecutivas en una lista y resumirlas en tuplas. Consigna:
Implementa compactar(xs) que devuelva una lista de tuplas (valor, cantidad consecutiva).


```python
def compactar(xs):
    if not xs:
        return []
    res = []
    actual, count = xs[0], 1
    for x in xs[1:]:
        if x == actual:
            count += 1
        else:
            res.append((actual, count))
            actual, count = x, 1
    res.append((actual, count))
    return res
```
