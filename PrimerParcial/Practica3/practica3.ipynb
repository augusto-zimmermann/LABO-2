```python
def sumar_ventas(ventas):
    """
    Recibe una lista de tuplas (producto, cantidad) y devuelve un diccionario
    con el total vendido por producto.
    """
    totales = {}
    for producto, cantidad in ventas:
        if producto in totales:
            totales[producto] += cantidad
        else:
            totales[producto] = cantidad
    return totales
```

```python
def palabras_largas(palabras, k):
    """
    Retorna las k palabras más largas de la lista.
    Si hay empates en longitud, mantiene el orden original.
    """
    indexed = list(enumerate(palabras))
    indexed_sorted = sorted(indexed, key=lambda ix_w: (-len(ix_w[1]), ix_w[0]))
    return [w for _, w in indexed_sorted[:k]]
```

## 4.

```python
def separar_pares_impares(nums):
    """
    Separa la lista de números en dos conjuntos: pares e impares.
    Devuelve (pares, impares).
    """
    pares = set()
    impares = set()
    for n in nums:
        if n % 2 == 0:
            pares.add(n)
        else:
            impares.add(n)
    return pares, impares
```

## 5.

```python
def promedio_mayores(nums, umbral):
    """
    Retorna el promedio de los números estrictamente mayores al umbral.
    Si no hay números, devuelve None.
    """
    mayores = [n for n in nums if n > umbral]
    if not mayores:
        return None
    return sum(mayores) / len(mayores)
```

### Consejo para el futuro con ciclos for

Preguntate qué necesitás producir con el ciclo.

- ¿Una lista? → usá append o list comprehension.

- ¿Una variable acumulada? → inicializala afuera y actualizala dentro.

- ¿Solo querés ejecutar algo por cada elemento? → no hace falta guardar nada.

- No dejes expresiones “flotando” como numero_grupo+1.
- Si no las guardás en una variable o no las usás en un print, Python las ignora.
